# Step 1: Import Libraries

 imports all required libraries for data handling, preprocessing, model training, and evaluation.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score
```

# Step 2: Load Datasets

In this step, we load the datasets manually from local files:
- IMDB Dataset
- SMS Spam Dataset

```python
# Load IMDB dataset
imdb = pd.read_csv("IMDB Dataset.csv")

# Load SMS dataset (your file name: spamraw.csv)
sms = pd.read_csv("spamraw.csv", encoding='latin-1')

# Keep only useful columns
sms = sms[['type', 'text']]
sms.columns = ['label', 'text']

print(imdb.head())
print(sms.head())
```

# Step 3: Data Preprocessing

We convert text labels into numeric values:
- IMDB: positive → 1, negative → 0
- SMS: spam → 1, ham → 0

```python
# IMDB labels
imdb['sentiment'] = imdb['sentiment'].map({'positive': 1, 'negative': 0})

# SMS labels
sms['label'] = sms['label'].map({'spam': 1, 'ham': 0})
```

# Step 4: Train-Test Split

We split the datasets into training and testing sets (80% training, 20% testing).

```python
# IMDB
X_train_imdb, X_test_imdb, y_train_imdb, y_test_imdb = train_test_split(
    imdb['review'], imdb['sentiment'], test_size=0.2, random_state=42
)

# SMS
X_train_sms, X_test_sms, y_train_sms, y_test_sms = train_test_split(
    sms['text'], sms['label'], test_size=0.2, random_state=42
)
```

# Step 5: Text Vectorization

We convert text into numerical features using TF-IDF.

```python
# IMDB vectorizer
vectorizer_imdb = TfidfVectorizer(stop_words='english', max_features=5000)

X_train_imdb_vec = vectorizer_imdb.fit_transform(X_train_imdb)
X_test_imdb_vec = vectorizer_imdb.transform(X_test_imdb)

# SMS vectorizer
vectorizer_sms = TfidfVectorizer(stop_words='english', max_features=5000)

X_train_sms_vec = vectorizer_sms.fit_transform(X_train_sms)
X_test_sms_vec = vectorizer_sms.transform(X_test_sms)
```

# Step 6: Logistic Regression

We train Logistic Regression on both datasets and evaluate accuracy.

```python
# IMDB
lr_imdb = LogisticRegression()
lr_imdb.fit(X_train_imdb_vec, y_train_imdb)

pred_imdb_lr = lr_imdb.predict(X_test_imdb_vec)
acc_imdb_lr = accuracy_score(y_test_imdb, pred_imdb_lr)

# SMS
lr_sms = LogisticRegression()
lr_sms.fit(X_train_sms_vec, y_train_sms)

pred_sms_lr = lr_sms.predict(X_test_sms_vec)
acc_sms_lr = accuracy_score(y_test_sms, pred_sms_lr)

print("Logistic Regression IMDB Accuracy:", acc_imdb_lr)
print("Logistic Regression SMS Accuracy:", acc_sms_lr)
```

# Step 7: Naive Bayes

We train Multinomial Naive Bayes on both datasets and evaluate accuracy.

```python
# IMDB
nb_imdb = MultinomialNB()
nb_imdb.fit(X_train_imdb_vec, y_train_imdb)

pred_imdb_nb = nb_imdb.predict(X_test_imdb_vec)
acc_imdb_nb = accuracy_score(y_test_imdb, pred_imdb_nb)

# SMS
nb_sms = MultinomialNB()
nb_sms.fit(X_train_sms_vec, y_train_sms)

pred_sms_nb = nb_sms.predict(X_test_sms_vec)
acc_sms_nb = accuracy_score(y_test_sms, pred_sms_nb)

print("Naive Bayes IMDB Accuracy:", acc_imdb_nb)
print("Naive Bayes SMS Accuracy:", acc_sms_nb)
```

# Step 8: Model Comparison

We compare all models based on accuracy.

```python
results = pd.DataFrame({
    'Model': [
        'Logistic Regression (IMDB)',
        'Naive Bayes (IMDB)',
        'Logistic Regression (SMS)',
        'Naive Bayes (SMS)'
    ],
    'Accuracy': [
        acc_imdb_lr,
        acc_imdb_nb,
        acc_sms_lr,
        acc_sms_nb
    ]
})

results = results.sort_values(by='Accuracy', ascending=False)
print(results)
```
